### Question 4


```python
# 计算 shot_made_flag 的值计数
shot_counts = kobe['shot_made_flag'].value_counts().sort_index()

px.bar(x=shot_counts.index, y=shot_counts.values, 
             labels={'x': 'Shot Made Flag', 'y': '频数'},
        title='Shot Made Flag 分布')
```

```python
total=len(kobe)
positive_ratio=shot_counts[1]/total
negative_ratio=shot_counts[0]/total

print(f"正样本比例: {positive_ratio:.2%}")
print(f"负样本比例: {negative_ratio:.2%}")
```

### Question 7


```python
data=kobe.groupby(['combined_shot_type', 'shot_made_flag']).size()
data
```

```python
data=kobe.groupby(['combined_shot_type', 'shot_made_flag']).size().unstack()
data
```

```python
grouped_data = kobe.groupby(['combined_shot_type', 'shot_made_flag']).size().unstack()

px.bar(grouped_data, 
             barmode='group', #并排参数而不是堆叠 stack
             labels={'value': '频数', 'combined_shot_type': '投篮类型', 'shot_made_flag': '投篮结果'},
             title='不同投篮类型的命中和未命中频数')
```
