```python
occupation1 = userinfo_df[["new_user_id","职业"]]
default1 = overdue_df[["new_user_id","违约"]]
weiyue_data = pd.merge(occupation1, default1, how="outer", on="new_user_id")
weiyue_data
```

```python
bank_df = bank_df[["new_user_id", "交易时间戳", "交易类型", "交易金额", "工资收入标记"]]
bank_df
```

```python
salary = bank_df[bank_df['工资收入标记'] == 1]
salary
```

```python
avg_salary=salary.groupby(['new_user_id'])["交易金额"].agg(["mean"])
avg_salary
```

```python
avg_salary = avg_salary.rename(columns={'mean': '平均收入'})
avg_salary.reset_index(inplace=True)
```

```python
merge = pd.merge(avg_salary, userinfo_df, on='new_user_id')
merge
```

```python
px.scatter(x=merge["教育程度"], y=merge["平均收入"])
```
