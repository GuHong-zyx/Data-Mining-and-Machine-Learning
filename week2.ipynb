```python
import os
import csv
```

```python
file_path = r"D:\WeChat Files\wxid_vzlr2hkxrgil22\FileStorage\File\2024-09\data3.csv"
non_smokers_file = r"D:\WeChat Files\wxid_vzlr2hkxrgil22\FileStorage\File\2024-09\2.csv"
smokers_file = r"D:\WeChat Files\wxid_vzlr2hkxrgil22\FileStorage\File\2024-09\1.csv"
bmi_index_file = r"D:\WeChat Files\wxid_vzlr2hkxrgil22\FileStorage\File\2024-09\3.csv"
smokers_age_file = r"D:\WeChat Files\wxid_vzlr2hkxrgil22\FileStorage\File\2024-09\4.csv"
```

```python
## 第一题
if os.path.exists(file_path):
    with open(file_path,mode='r',encoding='utf-8') as file:
        csv_reader=csv.reader(file)
        headers=next(csv_reader)
        with open(non_smokers_file,mode='w',newline='',encoding='utf-8') as non_smokers, \
             open(smokers_file,mode='w',newline='',encoding='utf-8') as smokers:
            non_smokers_writer=csv.writer(non_smokers)
            smokers_writer=csv.writer(smokers)
            smokers_writer.writerow(headers)
            non_smokers_writer.writerow(headers) 
            for row in csv_reader:
                if row[headers.index('smoke')] == '0':
                    non_smokers_writer.writerow(row)
                elif row[headers.index('smoke')] == '1':
                    smokers_writer.writerow(row)         
else:
    print('NO CSV！')
```

```python
## 第二题
if os.path.exists(file_path):
    with open(file=file_path,mode='r',newline='',encoding='utf-8') as file:
        file_reader=csv.DictReader(file)  # 以字典方式读取,便于后续
        count_male=0
        count_female=0
        male_age_sum=0
        female_age_sum=0
        male_bmi_sum=0
        female_bmi_sum=0   #初始化变量,因为不了解CSV模块的调用方式,采用了相对原始的方法
        for row in file_reader:
            age=float(row['age'])
            weight=float(row['weight'])
            height=float(row['height'])/100  # 为了方便后续调用变量名,命名后化为浮点数便于计算
            row['BMI']=weight/(height**2)   # 计算BMI指数
            bmi=row['BMI']
            if row['gender']=='1':
                male_age_sum+=age
                male_bmi_sum+=bmi
                count_male+=1
            elif row['gender']=='2':
                female_age_sum+=age
                female_bmi_sum+=bmi
                count_female+=1
        with open(file=bmi_index_file,mode='w',encoding='utf-8',newline='') as bmi_index:
            bmi_index_write=csv.writer(bmi_index)
            male_age_average=float(male_age_sum/count_male)
            female_age_average=float(female_age_sum/count_female)
            male_bmi_average=float(male_bmi_sum/count_male)
            female_bmi_average=float(female_bmi_sum/count_female)
            answer_male=[male_age_average,male_bmi_average]
            answer_female=[female_age_average,female_bmi_average]
            bmi_index_write.writerow(answer_male)
            bmi_index_write.writerow(answer_female)
            print(answer_male)
            print(answer_female)
else:
    print('NO CSV！')
```

```python
## 第三题
if os.path.exists(file_path):
    with open(file=file_path,mode='r',newline='',encoding='utf-8') as file:
        file_reader=csv.DictReader(file)  # 以字典方式读取,便于后续
        list_smoke=[]
        list_nosmoke=[]
        for row in file_reader:
            if row['smoke'] == '0':
                list_nosmoke.append(float(row['age']))
            elif row['smoke'] == '1':
                list_smoke.append(float(row['age']))
        age_smoke=statistics.median(list_smoke)/365
        age_nosmoke=statistics.median(list_nosmoke)/365
        age_difference=(statistics.median(list_nosmoke)-statistics.median(list_smoke))/365
        list_answer=[age_smoke,age_nosmoke,age_difference]
        print(list_answer)   
        with open(file=smokers_age_file,mode='w',newline='',encoding='utf-8') as smoke_file:
            smoke_file=csv.writer(smoke_file)
            for i in list_answer:
                smoke_file.writerow([i])  ##这里既可以写三个列表直接写入,也可以用一个小循环代替
else:
    print('NO CSV！')
```
