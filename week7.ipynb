#### Q1.利用梯度下降法求解$min_x(x-1)^2$的最优解

```python
def cost_function(x):
    return 2*(x-1)
```

```python
def gradient_descent(x0, alpha, N, epsilon):
    x=x0
    for i in range(N):
        grad=cost_function(x)
        x_new=x-alpha*grad
        if abs(x_new-x)<epsilon:
            break
        else:
            x=x_new
    return x
```

```python
gradient_descent(x0=0,alpha=0.05,N=200,epsilon=1e-6)
```
